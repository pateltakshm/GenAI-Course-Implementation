```python
from sklearn.preprocessing import OneHotEncoder
import numpy as np
```

```python
document =  ["My name is Taksh and I love AI and a dietitian"]
```

```python
documents = ["people watch movie", 
             "people watch cricket",
             "people like movie",
             "So I hate movies",
             "I am Taksh and I like a dietitian and AI field as a whole"]
```

```python
tokens= [sentence.lower().split() for sentence in documents]
```

```python
tokens
```

```python
all_words = [[word] for sentence in tokens for word in sentence]
```

This is the input format `Word2Vec` expects — a list of "sentences", where each "sentence" is a list of tokens. By wrapping individual words, each word becomes its own one-word sentence, which is a common approach when training on individual vocabulary items rather than full sentences.

```python
all_words
```

`sparse_output=False` means the encoder returns a NumPy array instead of a sparse matrix. This makes the output easier

```python
encoder = OneHotEncoder(sparse_output=False, handle_unknown= 'ignore')
```

```python
encoder.fit(all_words) 
```

```python
print("vocabulary: ", encoder.categories_)
```

```python
token= [word.lower().split() for word in document]
token
```

```python
for sentence in token:
    print(sentence)
```

```python
for sentence in token:
    encoded_sentence = encoder.transform([[word] for word in sentence])  
    print(f"\nSentence: {sentence}")
    print(encoded_sentence)
```

### BOW

`Bag of Words (BOW)` represents a document as a vector of word counts, ignoring order and grammar. And `Sklearn.CountVectorizer` does this in one step.

- **Ignores word order**: "dog bites man" = "man bites dog"                                         
- **No semantic meaning**: "movie" and "film" are treated as completely unrelated                      
- **Sparse vectors**: 10k vocab → mostly zeros per document                                             
- **Common words dominate**: "the", "is", "a" get "high counts despite carrying no meaning      

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python

documents = ["people watch movie and watch movie again", 
             "people watch cricket and watch cricket",
             "people like movie and like movie a lot",
             "people like cricket"]
```

```python
document = ["Taksh is a good learner but he struggles as soon as he distracts himself scrolling through social media"]
```

```python
bow = CountVectorizer()
```

```python
bow_matrix = bow.fit_transform(document)
```

```python
bow_matrix
```

```python
print("BOW Matrix:\n", bow_matrix.toarray())
```

```python
print("\nVocabulary", bow.get_feature_names_out())
```

```python
bow.fit(documents)
```

```python
bow.transform(documents).toarray()
```

### TF-IDF Practical

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
documents = ["people love to go out in the summer", 
             "people like to watch cricket matches in summer",
             "people like to drink cold beverage in summer",
             "people like to take vacation in summer"]
```

```python
tf_idf = TfidfVectorizer()
```

```python
tf_idf_vector = tf_idf.fit_transform(documents)
```

```python
tf_idf_vector.toarray()
```

```python
tf_idf.get_feature_names_out()
```

```python
print("vocabulary:", tf_idf.get_feature_names_out())
```
